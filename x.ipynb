```python
import pandas as pd

# Wczytanie danych ogólnych (płeć) i organizacji pracy (zarobki)
df_ogolne = pd.read_csv('0.df_ogolne.csv')
df_org = pd.read_csv('4.df_organizacja_pracy.csv')

# Połączenie tabel (zakładając wspólny identyfikator wiersza lub tę samą kolejność)
df_merged = pd.concat([df_ogolne[['płeć']], df_org[['zarobki']]], axis=1)

# Usunięcie braków danych i obliczenie średniej
srednie_zarobki = df_merged.groupby('płeć')['zarobki'].mean()
print(srednie_zarobki)

# Sprawdzenie różnicy
roznica = abs(srednie_zarobki.max() - srednie_zarobki.min())
if roznica < 200:
    print("Różnice są niewielkie (< 200 zł)")
else:
    print(f"Więcej zarabiają: {srednie_zarobki.idxmax()} (różnica: {roznica:.2f} zł)")
```
